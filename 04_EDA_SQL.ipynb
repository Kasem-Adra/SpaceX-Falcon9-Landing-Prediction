```
%load_ext sql
import pandas as pd
import sqlalchemy

df = pd.read_csv('dataset_part_2.csv')

# Create in-memory SQLite DB
engine = sqlalchemy.create_engine('sqlite:///spacex.db')
df.to_sql('SPACEXTBL', engine, if_exists='replace', index=False)
%sql sqlite:///spacex.db
print("Database ready!")
```

## Q1 – Display unique launch sites

```
%sql SELECT DISTINCT LaunchSite FROM SPACEXTBL;
```

## Q2 – 5 records where launch site begins with 'CCA'

```
%sql SELECT * FROM SPACEXTBL WHERE LaunchSite LIKE 'CCA%' LIMIT 5;
```

## Q3 – Total payload mass carried by NASA (CRS)

```
%sql SELECT SUM(PayloadMass) AS Total_Payload_NASA
FROM SPACEXTBL
WHERE Customer LIKE '%NASA (CRS)%';
```

## Q4 – Average payload mass by booster version F9 v1.1

```
%sql SELECT AVG(PayloadMass) AS Avg_Payload_v1_1
FROM SPACEXTBL
WHERE BoosterVersion LIKE '%v1.1%';
```

## Q5 – Date of first successful ground pad landing

```
%sql SELECT MIN(Date) AS First_Ground_Success
FROM SPACEXTBL
WHERE Outcome LIKE '%True RTLS%';
```

## Q6 – Boosters with successful drone ship landing & payload 4000–6000 kg

```
%sql SELECT BoosterVersion FROM SPACEXTBL
WHERE Outcome LIKE '%True ASDS%'
  AND PayloadMass BETWEEN 4000 AND 6000;
```

## Q7 – Total successful vs failed mission outcomes

```
%sql SELECT Class, COUNT(*) AS Count FROM SPACEXTBL GROUP BY Class;
```

## Q8 – Booster versions with max payload mass

```
%sql SELECT BoosterVersion, PayloadMass FROM SPACEXTBL
WHERE PayloadMass = (SELECT MAX(PayloadMass) FROM SPACEXTBL);
```

## Q9 – Failed drone ship landings in 2015

```
%sql SELECT BoosterVersion, LaunchSite, Outcome FROM SPACEXTBL
WHERE Outcome LIKE '%False ASDS%'
  AND Date LIKE '2015%';
```

## Q10 – Landing outcome rank between 2010-06-04 and 2017-03-20

```
%sql SELECT Outcome, COUNT(*) AS Count FROM SPACEXTBL
WHERE Date BETWEEN '2010-06-04' AND '2017-03-20'
GROUP BY Outcome
ORDER BY Count DESC;
```
