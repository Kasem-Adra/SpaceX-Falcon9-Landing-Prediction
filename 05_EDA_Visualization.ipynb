# 🚀 SpaceX Falcon 9 – Notebook 5: EDA with Visualization

Explore relationships using matplotlib and seaborn.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('dataset_part_2.csv')
print(df.shape)
df.head()
```

## 1 – Flight Number vs Launch Site

```
plt.figure(figsize=(10,5))
sns.scatterplot(x='FlightNumber', y='LaunchSite', hue='Class', data=df, palette={0:'red',1:'green'}, s=80)
plt.title('Flight Number vs Launch Site (Green = Success)')
plt.tight_layout()
plt.savefig('fig1_flightnum_launchsite.png', dpi=150)
plt.show()
```

## 2 – Payload Mass vs Launch Site

```
plt.figure(figsize=(10,5))
sns.scatterplot(x='PayloadMass', y='LaunchSite', hue='Class', data=df, palette={0:'red',1:'green'}, s=80)
plt.title('Payload Mass vs Launch Site')
plt.tight_layout()
plt.savefig('fig2_payload_launchsite.png', dpi=150)
plt.show()
```

## 3 – Success Rate per Orbit

```
orbit_success = df.groupby('Orbit')['Class'].mean().sort_values(ascending=False)
plt.figure(figsize=(12,5))
orbit_success.plot(kind='bar', color='steelblue')
plt.title('Average Success Rate per Orbit')
plt.ylabel('Success Rate')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('fig3_orbit_success.png', dpi=150)
plt.show()
```

## 4 – Flight Number vs Orbit Type

```
plt.figure(figsize=(10,5))
sns.scatterplot(x='FlightNumber', y='Orbit', hue='Class', data=df, palette={0:'red',1:'green'}, s=80)
plt.title('Flight Number vs Orbit')
plt.tight_layout()
plt.savefig('fig4_flightnum_orbit.png', dpi=150)
plt.show()
```

## 5 – Payload Mass vs Orbit

```
plt.figure(figsize=(10,5))
sns.scatterplot(x='PayloadMass', y='Orbit', hue='Class', data=df, palette={0:'red',1:'green'}, s=80)
plt.title('Payload Mass vs Orbit')
plt.tight_layout()
plt.savefig('fig5_payload_orbit.png', dpi=150)
plt.show()
```

## 6 – Success Rate over the Years

```
df['Year'] = pd.to_datetime(df['Date']).dt.year
yearly = df.groupby('Year')['Class'].mean()
plt.figure(figsize=(10,4))
yearly.plot(kind='line', marker='o', color='steelblue')
plt.title('Yearly Launch Success Rate')
plt.ylabel('Success Rate')
plt.xticks(yearly.index)
plt.grid(True)
plt.tight_layout()
plt.savefig('fig6_yearly_success.png', dpi=150)
plt.show()
```
