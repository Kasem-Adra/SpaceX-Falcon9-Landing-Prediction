# 🚀 SpaceX Falcon 9 – Notebook 3: Data Wrangling

Clean the dataset and create the binary **Class** label (1 = successful landing, 0 = failure).

```
import pandas as pd
import numpy as np

df = pd.read_csv('dataset_part_1.csv')
print("Shape:", df.shape)
df.head()
```

```
# Check missing values
print(df.isnull().sum())
```

```
# Unique landing outcomes
print(df['Outcome'].value_counts())
```

```
# Define bad outcomes
bad_outcomes = {'False ASDS', 'False RTLS', 'False Ocean', 'None ASDS', 'None None', 'None RTLS'}

# Create binary Class column
df['Class'] = df['Outcome'].apply(lambda x: 0 if x in bad_outcomes else 1)
print("Success rate:", df['Class'].mean().round(2))
df[['Outcome','Class']].head(20)
```

```
df.to_csv('dataset_part_2.csv', index=False)
print("Saved → dataset_part_2.csv")
print(df['Class'].value_counts())
```
